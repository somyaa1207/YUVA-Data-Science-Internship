```python
df.columns=df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
df.rename(columns={'socity' : 'society'}, inplace=True)
```

```python
df.columns
```

```python
df['price'].unique()[:20]
```

```python
df['price']=df['price'].apply(lambda x:x.strip())
```

```python
df['price'] = df['price'].str.strip().str.replace(',', '').astype(float)
```

```python
df['rate_per_sqft']=df['rate_per_sqft'].apply(lambda x:x.strip())
```

```python
df['rate_per_sqft'] = df['rate_per_sqft'].str.strip().str.replace(',', '').astype(float)
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df['status']=df['status'].str.lower().str.strip()

```

```python
df['rera_approval']=df['rera_approval'].str.lower().str.strip().map({'not approved by rera':False, 'approved by rera':True})
```

```python
categorical_cols = df.select_dtypes(include="string").columns

df[categorical_cols] = df[categorical_cols].apply(
    lambda col: col.str.lower().str.strip()
)
```

```python
df.describe(include='string')
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['price'],bins=50,kde=True)
plt.title('Distribution of Property Prices')
plt.xlabel('price')
plt.ylabel('No. of Properties')
plt.show()          
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df[df['area']<=10000]['area'], bins=50, kde=True)
plt.title('Distribution of Property Area (Up to 10,000 sq ft')
plt.xlabel('Area (sq ft)')
plt.ylabel('Number of Properties')
plt.xticks(range(0,10001,2000))
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='bhk_count', data=df)
plt.title('Distribution of BHK Count')
plt.xlabel('BHK Count')
plt.ylabel('Number of Properties')
plt.xticks(rotation=45)
plt.show()
```

# NOW EDA FOR CATEGORICAL COLUMNS/DATASET

```python
plt.figure(figsize=(6,4))
df['flat_type'].value_counts().plot(kind='bar')
plt.title('Distribution of Flat Type')
plt.xlabel('flat type')
plt.ylabel('No. of Properties')
plt.show()
```

```python
plt.figure(figsize=(6,4))
df['status'].value_counts().plot(kind='bar')
plt.title('Distribution of Flat Type')
plt.xlabel('flat type')
plt.ylabel('No. of Properties')
plt.show()
```

```python
plt.figure(figsize=(6,4))

status_counts = df['status'].value_counts()

plt.pie(
    status_counts,
    labels=None,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=1.15
)

plt.title('Property Status Distribution')

plt.legend(
    status_counts.index,
    title='Property Status',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))

rera_counts = df['rera_approval'].value_counts()

plt.pie(
    rera_counts,
    labels=rera_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Rera Approval Distribution')
plt.tight_layout()
plt.show()
```

```python
top_companies = df['company_name'].value_counts().head(10)

plt.figure(figsize=(8,6))

top_companies.plot(kind='bar')

plt.title('Top 10 Companies by Number of Properties')
plt.xlabel('Company Name')
plt.ylabel('Number of Properties')

plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

## Bivariate Analysis

Price vs Area

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='area',
    y='price',
    data=df
)
plt.title('Price vs Area')
plt.xlabel('area')
plt.ylabel('price')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='rate_per_sqft',
    y='price',
    data=df
)

plt.title('Property Price vs Rate per Sqft')
plt.xlabel('Rate per Sqft')
plt.ylabel('Price')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='bhk_count',
    y='price',
    data=df
)

plt.title('Property Price vs BHK Count')
plt.xlabel('BHK Count')
plt.ylabel('Price')

plt.tight_layout()
plt.show()
```

# Correlation Analysis

```python
correlation_matrix=df[['price', 'area', 'rate_per_sqft', 'bhk_count']].corr()
correlation_matrix
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```

 # Anomalies

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.boxplot(y=df['price'], ax=axes[0])
axes[0].set_title('Price — Outlier Analysis')
axes[0].set_ylabel('Price')

sns.boxplot(y=df['area'], ax=axes[1])
axes[1].set_title('Area — Outlier Analysis')
axes[1].set_ylabel('Area (sq ft)')

sns.boxplot(y=df['rate_per_sqft'], ax=axes[2])
axes[2].set_title('Rate per Sqft — Outlier Analysis')
axes[2].set_ylabel('Rate per Sqft')

plt.tight_layout()
plt.show()
```
